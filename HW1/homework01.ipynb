```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

iris_df = pd.read_csv("iris.data", header=None,
    names=["Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width", "Species"])
iris_attributes = iris_df[["Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width"]]

print(iris_df.head())
```

## 1. Compute STDev and Boxplots

```python
print(iris_attributes.std())

iris_attributes.boxplot()
plt.title("Boxplots of Iris Attributes")
plt.ylabel("Measurement (cm)")
plt.show()
```

## 2. Generate Histograms

```python
iris_attributes.hist(bins=10)
plt.suptitle("Histograms of Iris Attributes")
plt.show()
```

## 3. Generate QQ Plots for Sepal.Length and Sepal.Width

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

stats.probplot(iris_df["Sepal.Length"], dist="norm", plot=axes[0])
axes[0].set_title("QQ Plot: Sepal.Length")

stats.probplot(iris_df["Sepal.Width"], dist="norm", plot=axes[1])
axes[1].set_title("QQ Plot: Sepal.Width")

plt.tight_layout()
plt.show()
```

## 4.  Compute Scatter Plots between Sepal.Length and Sepal.Width and also Sepal.Length and Petal.Length

```python
plt.scatter(iris_df["Sepal.Length"], iris_df["Sepal.Width"])
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width")
plt.show()

print(f"Correlation: {iris_df["Sepal.Length"].corr(iris_df["Sepal.Width"])}")

plt.scatter(iris_df["Sepal.Length"], iris_df["Petal.Length"])
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.title("Sepal Length vs Petal Length")
plt.show()

print(f"Correlation: {iris_df["Sepal.Length"].corr(iris_df["Petal.Length"])}")
```
