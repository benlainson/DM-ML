```python
#import libraries
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#load data
df = pd.read_csv("titanic.csv")
df.head()
```

```python
#convert columns to categories
data = pd.DataFrame()

data["Survived"] = df["Survived"].map({1: "Yes", 0: "No"})
data["Class"]    = df["Pclass"].astype(str)
data["Sex"]      = df["Sex"]

data["Age"] = pd.cut(df["Age"],
                     bins=[0, 18, 30, 50, 100],
                     labels=["Child", "YoungAdult", "Adult", "Senior"],
                     right=False)

data["Fare"] = pd.cut(df["Fare"],
                      bins=[0, 10, 30, 100, 1000],
                      labels=["Low", "Mid", "High", "VeryHigh"],
                      right=False)

data["SibSp"] = (df["Siblings/Spouses Aboard"] > 0).map({True: "1+", False: "0"})
data["ParCh"] = (df["Parents/Children Aboard"] > 0).map({True: "1+", False: "0"})

data.head()
```

```python
#one-hot encode
basket = pd.get_dummies(data).astype(bool)
basket.head()
```

```python
#setting min support and min confidence with 0.2/0.5
frequent1 = apriori(basket, min_support=0.20, use_colnames=True)
rules1 = association_rules(frequent1, metric="confidence", min_threshold=0.50)

print("Frequent itemsets:", len(frequent1))
print("Rules:", len(rules1))
rules1[["antecedents", "consequents", "support", "confidence", "lift", "leverage", "conviction"]].sort_values("confidence", ascending=False).head(10)
```

```python
#setting combination 2 with 0.1/0.7
frequent2 = apriori(basket, min_support=0.10, use_colnames=True)
rules2 = association_rules(frequent2, metric="confidence", min_threshold=0.70)

print("Frequent itemsets:", len(frequent2))
print("Rules:", len(rules2))
rules2[["antecedents", "consequents", "support", "confidence", "lift", "leverage", "conviction"]].sort_values("confidence", ascending=False).head(10)
```

```python
#setting combination 3 with 0.05/0.8
frequent3 = apriori(basket, min_support=0.05, use_colnames=True)
rules3 = association_rules(frequent3, metric="confidence", min_threshold=0.80)

print("Frequent itemsets:", len(frequent3))
print("Rules:", len(rules3))
rules3[["antecedents", "consequents", "support", "confidence", "lift", "leverage", "conviction"]].sort_values("confidence", ascending=False).head(10)
```

```python
#rank the top 5 association rules by confidence
rules3.sort_values(["confidence", "lift"], ascending=False).head(5)
```

##### The confidence rule highlights that many rules tie at confidence 1, so lifting breaks the ties.

```python
#ranking top 5 association by lift
rules3.sort_values("lift", ascending=False).head(5)
```
